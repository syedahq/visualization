```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Making a df, using the Open Data Website for Deaths of People Experiencing Homelessness organized by month and year
# The data contains the year of death, month of death, and total number of deaths
data = { "Year of death": [2022, 2023, 2024, 2023, 2023, 2023, 2024, 2024, 2024, 2022, 2023, 2022, 2022, 2023, 2024, 2022, 2023, 2022, 2022, 2024, 2023, 2022, 2024, 2023, 2022, 2024, 2023, 2022, 2023, 2022, 2024, 2022, 2024, 2023, 2024, 2024],

"Months": ["July","February","February","April","March","November","March","December","October","May","June","April","March","July","November","February","May","December","November","April","October","January","September","January","September","May","August","August","December","October","June","June", "July","September","January","August"],

"Death_Count": [33,20,18,16,23,35,19,13,13,32,25,26,22,29,12,18,20,33,33,25, 26,35,10,26,18,24,20,27,26,21,14,21,13,32,34,16] }

# Convert the dictionary into a Pandas DataFrame
df = pd.DataFrame(data)

#I only want to plot data from the year 2022, so I am filtering that
# Filtering the dataset allows the chart to focus on one year rather than combining multiple years together
df_2022 = df[df["Year of death"] == 2022]

# For this visualization, I only want to examine deaths occurring during 2022. Filtering the dataset allows the chart to focus on one year rather than combining multiple years together
month_order = ["January","February","March","April","May","June","July","August","September","October","November","December"]

df_2022["Months"] = pd.Categorical(df_2022["Months"], categories=month_order, ordered=True)

## Increasing the figure size improves readability and prevents labels from appearing crowded
plt.figure(figsize=(14,6))

#A bar chart is appropriate because the goal is to compare the number of deaths across discrete categories (months)
#A single color is used to reduce unnecessary visual distractions 
sns.barplot(
    data=df_2022,
    x="Months",
    y="Death_Count",
    ci=None,
    color="red"
)

#Rotating labels prevents month names from overlapping
plt.xticks(rotation=45, ha='right')

#Forming the title of the axis and main title
plt.title("Homeless deaths by month in Toronto in the year 2022")
plt.xlabel("Month")
plt.ylabel("Number of deaths")

#tight_layout() automatically adjusts spacing so that titles, labels, and tick marks fit properly within the figure boundaries
plt.tight_layout()

#display the visualization
plt.show()
```

    > What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

    Pupose and perception: Bar chart was chosen as the data involved comparing values across categorical groups (months). According to the visualization selection guides such as the Financial Times Visual Vocabulary, bar charts are effective for comparisons tasks. 

    Reducing cognitive load: To reducde extraneous cognitive load, I ordered the months chornologically rather than than alphabetically, I used a commonly used chart type, I kept it minimilistic by using essential visual elements and I chose a single color to avoid unncessary complexity. 

    Readability: I increased the size of the figure, rotated month labels to prevent overlap, used clear axis labels and a descriptive title to help viewers immediately understand what is being plotted. 

    Perceived factual basis: In class, we learned that 2D layouts, clean designs and geometric shapes tend to enhance perception of objectivity. My plot follows these conventions by using a simple 2D bar chart without decorative elements or 3D effects. 

    Gestalt principles: That bars are grouped within a frame, they share the same color, enabling viewers to perceive them as part of a single dataset and focus on differences in height.
